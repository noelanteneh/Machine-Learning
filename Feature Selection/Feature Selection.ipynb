```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('cardio_train.csv',sep=';')
```

```python
df.head()
```

```python
x = df.drop(['cardio', 'id'],axis=1)
x
```

```python
y=df.cardio
y
```

```python
from sklearn.feature_selection import SelectKBest
```

###  1. f_classif (target variable is categorical and your input features are numerical)

```python
from sklearn.feature_selection import f_classif
```

```python
selector=SelectKBest(score_func=f_classif)
```

```python
selector.fit(x,y)
```

```python
selector.scores_
```

```python
score_col = pd.DataFrame(data=selector.scores_, columns=['score'])
score_col 
```

```python
Name_col=pd.DataFrame(data=x.columns,columns=['selector'])
Name_col
```

```python
selector_score=pd.concat([Name_col,score_col], axis=1)
selector_score
```

```python
selector_score.nlargest(8,'score')
```

### 2.Chi2 (for non- negative inputs)

```python
from sklearn.feature_selection import chi2

selector = SelectKBest(score_func=chi2)
```

```python
# Find all negative values in x
(x < 0).sum()
```

```python
valid = (x['ap_hi'] >= 0) & (x['ap_lo'] >= 0)

x = x.loc[valid]
y = y.loc[valid]
```

```python
selector.fit(x,y)
```

```python
score_col = pd.DataFrame(data=selector.scores_, columns=['score'])
Name_col = pd.DataFrame(data=x.columns, columns=['selector'])
selector_score = pd.concat([Name_col, score_col], axis=1)
selector_score.nlargest(8, 'score')
```

### 3.mutual_info_classif(including nonlinear relationships)

```python
from sklearn.feature_selection import mutual_info_classif
```

```python
selector = SelectKBest(score_func=mutual_info_classif)
```

```python
selector.fit(x,y)
```

```python
score_col = pd.DataFrame(data=selector.scores_, columns=['score'])
Name_col = pd.DataFrame(data=x.columns, columns=['selector'])
selector_score = pd.concat([Name_col, score_col], axis=1)
selector_score.nlargest(8, 'score')
```
