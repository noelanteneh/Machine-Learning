### Reorganizing Arrays

```python
before= np.array([[1,2,3,4],[5,6,7,8]])
print (before)
print ("#########################")

after= before.reshape((4,2))
print (after)
```

```python
#vertically stacking vectors 
v1=np.array([1,2,3,4])
v2=np.array([2,3,4,5])
np.vstack([v1,v2])
```

```python
# horizontal tacking vectors 
h1=np.array([1,2,3,4])
h2=np.array([2,3,4,5])
np.hstack([v1,v2])
```

```python
def fizzbuzz(n):
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

# Example
fizzbuzz(15)
```
