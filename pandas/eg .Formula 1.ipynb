```python
import pandas as pd
import fastf1
import os

CACHE_DIR = r"C:\Users\noela\OneDrive\Desktop\Machine Learning\fastf1 cache"

os.makedirs(CACHE_DIR, exist_ok=True)

fastf1.Cache.enable_cache(CACHE_DIR)

print("FastF1 cache enabled!")
```

```python
session = fastf1.get_session(
    2025,
    "Las Vegas",
    "R"
)
```

```python
session.load()
```

```python
results = session.results

results.head()
```

```python
laps = session.laps

laps.head()
```

```python
hamilton = laps.pick_drivers("HAM")

hamilton.head()
```

```python
fastest = hamilton.pick_fastest()

fastest
```

```python
telemetry = fastest.get_car_data()

telemetry.head()
```

```python
lap_data = laps[
    [
        "Driver",
        "LapNumber",
        "LapTime",
        "Compound",
        "TyreLife",
        "FreshTyre",
        "Position"
    ]
]
```

```python
lap_data.head()
```

```python
lap_data.isnull().sum()
```

```python
lap_data = lap_data.dropna()
```

```python
lap_data.isnull().sum()
```

```python
lap_data["LapTime"] = lap_data["LapTime"].dt.total_seconds()
```

```python
lap_data.head()
```

```python
driver_average = (
    lap_data
    .groupby("Driver")["LapTime"]
    .mean()
)
```

```python
driver_average
```

```python
tyres = (
    lap_data
    .groupby(["Driver","Compound"])
    .size()
)

tyres
```

```python
lap_data.to_csv(
    "F1_lap_data.csv",
    index=False
)
```
