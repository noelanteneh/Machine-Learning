```python
def convert_value(value):
    if pd.isna(value):
        return None

    value = value.replace("€", "")

    if value.endswith("M"):
        return float(value[:-1]) * 1_000_000

    elif value.endswith("K"):
        return float(value[:-1]) * 1_000

    else:
        return float(value)
```

```python
df["Value"] = df["Value"].apply(convert_value)
```

```python
plt.figure(figsize=(10,7))

plt.scatter(
    df["Age"],
    df["Overall"],
    s=df["Value"]/1_000_000,   # Bubble size
    alpha=0.4
)

plt.xlabel("Age")
plt.ylabel("Overall")
plt.title("Age vs Overall ,Bubble Size = Market Value")
plt.grid(alpha=0.3)

plt.show()
```
