```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import numpy as np
```

```python
x=np.linspace(0,5,11)
y=x**2
```

```python
x
```

```python
y
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,'r')
plt.show()
```

```python
plt.plot(x,y,'r')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Graph')
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'g')
plt.subplot(1,2,2)
plt.plot(x,y,'b')
```

```python
## oo method
fig= plt.figure()
axes=fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,y)
```

```python
## oo method
fig= plt.figure()
axes=fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,y)
axes.set_xlabel('x axis')
axes.set_ylabel('y axis')
axes.set_title('title')
```

```python
fig=plt.figure()
axes1=fig.add_axes([0.1,0.1,0.8,0.8])
axes2=fig.add_axes([0.2,0.5,0.4,0.3])
```

```python
fig=plt.figure()
axes1=fig.add_axes([0.1,0.1,0.8,0.8])
axes2=fig.add_axes([0.2,0.5,0.4,0.3])
axes1.plot(x,y)
axes2.plot(y,x)
```

```python
fig,axes= plt.subplots(ncols=3,nrows=3)
plt.tight_layout()
```

```python
fig,axes= plt.subplots(ncols=2,nrows=1)
for curn in axes:
    curn.plot(x,y)


```

```python
fig,axes= plt.subplots(ncols=2,nrows=1)
axes[0].plot(x,y)
axes[0].set_title('first')
axes[1].plot(y,x)
axes[1].set_title('second')
plt.tight_layout()
```

#### Figure Size and DPI

```python
fig= plt.figure(figsize=(8,3))
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y)
```

```python
fig= plt.figure(figsize=(8,3))
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y, label='x,y')
axes.plot(y,x, label='x,y')
axes.legend(loc=0)
```

#### plot Appearance

```python
fig=plt.figure()
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y,color='green',linewidth=30,alpha=0.5)
```

```python
fig=plt.figure()
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y,color='red',linestyle='-.')
```

```python
fig=plt.figure()
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y,color='red',linestyle=':')
```

```python
fig=plt.figure()
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y,color='red',marker='*', markersize=10)
```

```python
fig=plt.figure()
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y,color='red',marker='o', markersize=10, markerfacecolor='yellow', markeredgecolor='blue' , markeredgewidth=3)
```
